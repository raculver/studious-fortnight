## Ole Peter's "Equation of life"
An example of non-ergodicity

At any point, the expectation value is for the system to gain value, but the overall trend is that everyone loses value

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a,b = [1.5, 0.6]

def eqol(N):
    xs = np.empty(N)
    xs[0] = 1
    for n in range(1,N):
        xs[n] = xs[n-1] * (a if (np.random.random() > 0.5) else b)
    return xs
```

```python
nRuns = 2000
N = 300
ts = np.arange(N)

xs = np.array([eqol(N) for i in range(nRuns)]).T
av = np.mean(xs, axis = 1)

ens_average = np.empty(N)
ens_average[0] = 1
for n in range(1,N):
    ens_average[n] = ens_average[n-1] * np.mean([a,b])
                                               
_ = plt.semilogy(ts, xs[:, :],'-', linewidth = 1)
_ = plt.semilogy(ts, ens_average,'-', linewidth = 2)
```
