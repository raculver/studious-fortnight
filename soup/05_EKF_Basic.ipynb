
# Extended Kalman Filter
- Drawbacks of KF: assumption that transition and sensor models be linear-Gaussian.
- "Linear" means the state and mesurements are linear functions of the state vector

\begin{align}\mathbf{x}_{k|k-1} &= F_{k} \mathbf{x}_{k-1} + B_{k}\mathbf{u}_{k}\\
      \mathbf{z}_{k|k-1} &= H_{k} \mathbf{x}_{k|k-1}\end{align}

- previously, we were calculating the mean expected position?... (check)

\begin{align}\mathbf{x}_{k|k−1} = \boldsymbol{\mu}_{k|k−1}.\end{align}

- in EKF we instead do a Taylor expansion
\begin{align}g(\mathbf{x})\rvert_{\mathbf{x}=\boldsymbol{\mu}} \approx \sum\limits_{|\alpha| \ge 0}
      \frac{(\mathbf{x} - \boldsymbol{\mu})^{\alpha}}{\alpha !}
      (\mathcal{D}^{\alpha} g)(\boldsymbol{\mu})\end{align}

- where $\mathcal{D}^{\alpha}g$ are the derivatives of $g$.

- previously
\begin{align} F (x_{k-1})\approx \frac{\partial g}{\partial x}\bigg\rvert_{x=\mu_{k-1}} \mbox{(linear case)}\end{align}


## Rob's notes

- EKF uses:
  - Nonlinear function $f$ to predict state
  - Local linear approx to propagate uncertainty and calc meas update
- This is why EKF equations look very similar to usual EKF
- Algos with cosine and angular rates are obviously not linear in the same way... not a linear, scaled sum of the state vector elements

## From $F$ to $f$

\begin{align}\mathbf{x}_{k|k-1} &= F_{k} \mathbf{x}_{k-1} + w_{k}\end{align}

becomes

\begin{align}\mathbf{x}_{k|k-1} &= f_{k} (\mathbf{x}_{k-1}) + w_{k}\end{align}

where $f(x)$ is a function rather than a matrix multiplication.

## Step 1: Predict the current state

Our best guess of the current state is:

\begin{align} \mathbf{\mu}_{k|k-1} = f(\mathbf{\mu}_{k-1}) \end{align}

Where $\mathbf{\mu}_{k-1}$ was the last best estimate. The transformation model updates this.

## Step 2: Predict uncertainty on this estimate

This was weird to me for ages, but essentially, instead of predicting the uncertainty on the state $x$, we predict the uncertainty on the deviation from the mean $x-\mu$. These are equivalent because when calculating the covariance $Q$, anything constant (the mean value, basically) will cancel out.

Previously, we would have done:

\begin{align} P_{k|k-1} = F P_{k-1} F^{T} + Q \end{align}

But... this requires that we have linear transformations!

We can approximate a linear transformation of $f(x)$ if we do a Taylor Series

\begin{align} f(x) \approx f(\mu) + \frac{\partial f}{\partial x}\bigg\rvert_{x=\mu}(x-\mu) \end{align}

The Jacobian gives us a matrix which behaves like

\begin{align} A = J(f(x))\bigg\rvert_{x=\mu} = \frac{\partial f}{\partial x}\bigg\rvert_{x=\mu} \end{align}

so

\begin{align} f(x) \approx f(\mu) + A(x-\mu) \end{align}

Which is basically a linear transition model.

We now say that

\begin{align} \tilde{F_k} \approx A = J(f)\bigg\rvert_{x=\mu_{k-1}}\end{align}

## What about the constants that we're completely ignoring?... we can't just willy nilly say $\tilde{F_k} \approx A$, this transition model looks nothing like the original transition model!

As mentioned above, we're calculating the covariance of the deviation from the mean... mean of $f(x)$ will be $f(\mu)$.

\begin{align} \mbox{Transition model} = f(x) - f(\mu) \approx A(x-\mu) \end{align}

## Should you be frightened and confused by the Jacobian?

No. It's really just a table of partial derivatives

\begin{align} f(x,y) = \begin{bmatrix}
                    x^2 + y \\
                    xy \\
                    \end{bmatrix} \\ \end{align}

\begin{align} J(f) = \begin{bmatrix}
                    2 x & 1 \\
                    y & x \\
                    \end{bmatrix} \\ \end{align}
